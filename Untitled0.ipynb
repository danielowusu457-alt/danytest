Simple VAT Calculator App



```python
# Install Gradio
!pip install gradio -q

import gradio as gr
```

```python


VAT_RATE = 0.20   # 20%

# Function for VAT calculations
def vat_calculator(amount, calculation_type):

    if amount <= 0:
        return "Please enter a valid amount."

    # Add VAT
    if calculation_type == "Add VAT":
        vat = amount * VAT_RATE
        total = amount + vat

        return (
            f"Original Amount: GHS {amount:,.2f}\n"
            f"VAT @20%: GHS {vat:,.2f}\n"
            f"Total Amount: GHS {total:,.2f}"
        )

    # Remove VAT
    elif calculation_type == "Remove VAT":
        vat_exclusive = amount / (1 + VAT_RATE)
        vat = amount - vat_exclusive

        return (
            f"VAT Inclusive Amount: GHS {amount:,.2f}\n"
            f"VAT Portion: GHS {vat:,.2f}\n"
            f"VAT Exclusive Amount: GHS {vat_exclusive:,.2f}"
        )

# Gradio Interface
app = gr.Interface(
    fn=vat_calculator,
    inputs=[
        gr.Number(label="Enter Amount (GHS)"),
        gr.Radio(
            ["Add VAT", "Remove VAT"],
            label="Select Calculation Type"
        )
    ],
    outputs=gr.Textbox(label="VAT Calculation Result"),

    title="Ghana VAT Calculator App",
    description="Simple VAT Calculator using 20% Effective VAT Rate",

    theme="soft"
)

# Launch App
app.launch(share=True)
```

```python
import gradio as gr

# VAT Rate
VAT_RATE = 0.20  # 20%

# VAT Calculation Function
def vat_calculator(amount, calculation_type):

    # Validation
    if amount <= 0:
        return "Please enter a valid amount greater than zero."

    # ADD VAT
    if calculation_type == "Add VAT":
        vat = amount * VAT_RATE
        total = amount + vat

        result = f"""
VAT EXCLUSIVE AMOUNT: GHS {amount:,.2f}

VAT @20%: GHS {vat:,.2f}

TOTAL VAT INCLUSIVE AMOUNT: GHS {total:,.2f}
"""
        return result

    # REMOVE VAT
    elif calculation_type == "Remove VAT":
        vat_exclusive = amount / (1 + VAT_RATE)
        vat = amount - vat_exclusive

        result = f"""
VAT INCLUSIVE AMOUNT: GHS {amount:,.2f}

VAT PORTION: GHS {vat:,.2f}

VAT EXCLUSIVE AMOUNT: GHS {vat_exclusive:,.2f}
"""
        return result


# Gradio Interface
app = gr.Interface(
    fn=vat_calculator,

    inputs=[
        gr.Number(
            label="Enter Amount (GHS)",
            value=100
        ),

        gr.Radio(
            choices=["Add VAT", "Remove VAT"],
            label="Choose Calculation Type",
            value="Add VAT"
        )
    ],

    outputs=gr.Textbox(
        label="VAT Calculation Result",
        lines=10
    ),

    title="Ghana VAT Calculator",
    description="Simple VAT Calculator using the 20% Effective VAT Rate",

    theme="soft"
)

# Launch App
app.launch()
```

```python


# Store tasks
tasks = []

# Add task function
def add_task(task):
    if task.strip() == "":
        return tasks, ""

    tasks.append("✅ " + task)
    return "\n".join(tasks), ""

# Clear tasks function
def clear_tasks():
    tasks.clear()
    return ""

# Custom CSS
custom_css = """
body {
    background: linear-gradient(to right, #141e30, #243b55);
}

.gradio-container {
    font-family: Arial, sans-serif;
}

h1 {
    text-align: center;
    color: white;
    font-size: 40px;
}

.task-box textarea {
    background-color: #ffffff !important;
    color: #000000 !important;
    border-radius: 15px !important;
    padding: 15px !important;
    font-size: 18px !important;
}

button {
    border-radius: 12px !important;
    font-size: 16px !important;
    font-weight: bold !important;
    padding: 10px !important;
}

footer {
    visibility: hidden;
}
"""

# Build Interface
with gr.Blocks(css=custom_css, theme=gr.themes.Soft()) as app:

    gr.Markdown(
        """
        # 🌟 Daily To-Do App
        ### Organize Your Day Beautifully
        """
    )

    with gr.Row():
        task_input = gr.Textbox(
            label="Enter Your Task",
            placeholder="Type your daily task here..."
        )

    with gr.Row():
        add_btn = gr.Button("➕ Add Task")
        clear_btn = gr.Button("🗑️ Clear Tasks")

    task_output = gr.Textbox(
        label="Your Daily Tasks",
        lines=12,
        elem_classes="task-box"
    )

    # Button actions
    add_btn.click(
        fn=add_task,
        inputs=task_input,
        outputs=[task_output, task_input]
    )

    clear_btn.click(
        fn=clear_tasks,
        outputs=task_output
    )

# Launch App
app.launch(share=True)
```
